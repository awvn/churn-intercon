### Projeto Final: Relatório da Solução

#### Passos executados vs Passos pulados
- Este projeto teve um plano de trabalho detalhado, com vários passos. A maior parte dos passos foram executados com excessão da otimização do hiperparâmetros do XGBoost (utilizando por ex: GridSearch ou similar), porque após fazer o balanceamento dos dados, fez-se o treinamento do modelo com parâmetros padrão e o mesmo passou a meta definida de 0.85, obtendo 0.9084 uma meta muito boa.

#### Dificuldades encontradas e como foram ultrapassadas
- Encontrei algumas dificuldades ao longo do projeto das quais passo a mencionar:
  - Aparição de dados nulos após junção de todas as tabelas em uma só: Utilizaram-se algumas técnicas como substituição de dados utilizando 'No', preenchimento de dados ausentes com '0' e conversão dos dados em numérico.
  - Desbalanceamento de dados, que poderia se tornar um perigo para o meu modelo: Não foi necessário ultrapassar o desbalanceamente, porque atingiu-se a meta exigida.
  - Colunas não categóricas: Este foi outro problema que levou algum tempo a ser resolvida, porque a resolução não foi uniforme. Algumas que já estavam em formato 'Yes' e 'No', foram agrupadas em uma lista e convertidas para 'Yes'=1 & 'No'=0. A coluna Gender migrou para o formato 1 e 0, e outras para 1 e 0 também. A coluna 'status' foi definida como o target.
- **dados:** 4 tabelas, 7.043 clientes e a taxa de cancelamento observada (cerca de 26,5%);
- **resultado além do AUC-ROC:** acurácia de 86,23% e a leitura da matriz de confusão que você fez no notebook (233 cancelamentos encontrados e 141 que passaram despercebidos);
- **recomendação para o negócio:** o marketing deve usar o modelopara aplicar as ofertas aos clientes com maior probabilidade de cancelamento e, se o custo de perder um cliente for alto, baixar o limiar de decisão para encontrar mais casos;
- **limitações e próximos passos:** validação cruzada para confirmar a estabilidade do resultado foi feita e o ajuste de hiperparâmetros que ficou de fora.

#### Etapas principais da resolução:
- A decisão de utilização de conversão directa para binários, e one-hot-enconding com algumas colunas categóricas com mais de duas opções, para a preparação dos dados para a criação dos modelos
- A Junção das 4 tabelas em 1
- A resolução do desbalanceamento dos dados
- Estas foram as etapas principais, porque sem os dados estarem conforme, seria difícil ou mesmo impossível a criação dos modelos e quiçá a resolução das questões colocadas para este projeto.

#### O Modelo final:
- O modelo final acabou por ser o XGBoost, e o mesmo teve uma pontuação de 0.9084 comparando os outros dois modelos (Logistic Regression e Random Forest, ambos com 0.84 de pontuação).
