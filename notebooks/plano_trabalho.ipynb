## Plano de Trabalho do Projecto
### Previsão de Churn - Interconnect

#### O plano de trabalho do projecto final do boot camp de DS, está constituído por 4 etapas que passo a numera-las:
 1. Objectivo do negócio e definição do problema;
 2. Visão geral das tabelas e cruzamento de Dados;
 3. Etapas de execução;
 4. Alinhamento de Negócio. 


#### ** Desdobramento das etapas **

#### 1. Objectivo do negócio e definição do problema

Contextualizando: A empresa de telecomunicações "Interconnect", enfrenta sérios problemas de perda de clientes, e pretende mitigar os problemas, antecipando o processo de cancelamento de contractos(churn).

O problema: É importante ter em conta que reter um cliente ativo é muito melhor e tem mais vantagens financeiras do que adiquirir um novo. Para se antecipar a "churn", a equipa de marketing necessita identificar os perfis propensos a deixarem a empresa para se oferecer pacotes especiais e planos promocionais.

Em termos técnicos: Pretende-se contruir um modelo de classificação binária para prever se o cliente irá cancelar o serviço (target: 1) ou continuar ativo (target: 0).

As métricas: O sucesso do projecto será rigirosamente avaliado pela métrica AUC-ROC 0.85. A métrica secundária será a Acurácia.



#### 2. Visão geral das tabelas e cruzamento de Dados

Com o intuito de se começar a análise exploratória dos dados (EDA), os dados brutos armazenados em 4 fontes diferentes precisam ser unificados:
- contract.csv: Informações contractuais do cliente
- pesonal.csv: Dados pessoais do cliente
- internet.csv: Detalhes sobre os serviços de internet contractados
- phone.csv: Detalhes sobre os serviços telefónicos contractados

Chave de cruzamento: Todas as tabelas têm em comum a coluna identificadora de cliente "customerID"



- A tabela contract tem 7043 linhas e 8 colunas, sem qualquer valor nulo, possui tipos de dados object e float;
- A tabela personal tem 7043 linhas e 5 colunas, sem qualquer valor nulo, possui tipos de dados object e int64;
- A tabela internet tem 5517 linhas e 8 colunas, sem qualquer valor nulo, possui tipos de dados object; 
- A tabela phone tem 6361 linhas e 2 colunas, sem qualquer valor nulo, possui tipos de dados object.

#### 3. Etapas de execução

Primeiro: Consolidação da variável alvo (Target) e Fusão  de Dados
- Cruzar as quatro tabelas usando o customerID.
- Criar a coluna Target com base na coluna EndDate. Caso EndDate = "No", o cliente continua como ativo(0) na base de dados. Se tiver um adata de rescisão de contracto, o cliente cancelou o serviço (1), não está ativo.

Segundo: Pré processamento e limpeza de dados (Data Prep & Cleaning)
- Ajuste dos tipos de dados: Conversão da coluna TotalCharges para float. Podem aparecer clientes em espaços brancos nesta coluna, e os mesmos devem ser tratados.
- Tratar os valores nulos: Clientes ausentes das tabelas internet.csv e phone.csv, significa que os mesmos não contrataram os serviços em questão, desta feita os espaços em branco devem ser preenchido com "No".
- Tempo de vida do contracto: O calculo de dias que o cliente permaneceu ou permanece na empresa é feito subtraindo BeginDate de EndDate ou usando a data de corte do dataset para contractos ativos.
- Uma vez definido a variável alvo e o tempo de contracto, é importante efectuar o drop das colunas BeginDate e EndDate, para não serem utilizadas pelo modelo. Quando o EndDate tem data; significa que o cliente cancelou, o que pode provocar métricas artificiais altas do modelo (vazamento de dados).

Terceiro: Engenharia de Recursos
- Lidar com as variáveis categóricas utilizando a codificação One-Hot-Encoding para modelos lineares ou permanecer com o formato em questão para árvores de decisão.
- Verificar e registrar o desbalanceamento das classes, porque podem viciar o modelo (por vezes tende a existir mais clientes ativos do que cancelados).
- A regressão logídtica, costuma funcionar melhor com variáveis numéricas namesma escala, desta feita é necessária efectuar um escalonamento.

Quarto: Treinamento, Validação e Ajuste dos Hiperparâmetros 
- Divisão dos dados em treino e teste (80/20 ou 75/25, mas vou por 80/20 por já estar habituado), utilizando estratificação (estratify=target) para manter a proporção das classes.
- Treinar os modelos base(Regressão Logística e Random Forest).
- Implementar modelos avançados de Gradient Boosting (LightGBM, CatBoost ou XGBoost)
- Optimização de hiperparâmetros (taxa de aprendizado, profundidade das alvores) utilizando GridSearchCV ou RandomizedSearchCV.
- Avaliar o modelo  final no conjunto  de teste usando curva ROC e a matriz de confusão.


#### 4. Perguntas Críticas: Alinhamento de Negócio
- Como tratar os valores nulos resultantes do merge (zerar, remover ou criar categoria)
- Como lidar com as colunas EndDate e BeginDate, ambas são do tipo object. Devem ser convertudas para tipo data, o "No" tem outro tratamento?
- Como lidar com os clientes com menos de 1 mês de contracto que estão  com a coluna TotalCharges em branco devem permanecer no dataset com valor zero, ou a falta de historico financeiro pode distorcer as previsões de churn a longo prazo?
